```python
import os
import cv2
from IPython.display import display
from PIL import Image
from ultralytics import YOLO
from ultralytics.utils import ASSETS
from detect import DEFAULT_MODEL, detect_frame, run_stream

# Muat model sekali agar bisa dipakai ulang di cell pengujian gambar.
model = YOLO(DEFAULT_MODEL)
```

## Uji Deteksi pada Gambar
Ganti `custom_image` ke file gambar lokal jika ingin menguji gambar tertentu.

```python
custom_image = "sample.jpg"
image_path = custom_image if os.path.exists(custom_image) else str(ASSETS / "bus.jpg")

frame = cv2.imread(image_path)
annotated = detect_frame(model, frame)
# Konversi BGR (OpenCV) ke RGB agar warna tampil benar di notebook.
display(Image.fromarray(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB)))
```
